# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
paragraph = """
Artificial intelligence is transforming industries across the globe. By automating routine 
tasks and providing powerful analytical insights, AI allows organizations to operate more 
efficiently. However, it also introduces ethical challenges such as data privacy concerns 
and algorithmic bias that require careful regulation.
"""

prompt = f"Summarize: {paragraph.strip()}"

# Tokenize and generate summary
inputs = tokenizer(prompt, return_tensors="pt")
summary_ids = model.generate(**inputs, max_new_tokens=50)

# Decode tokens back to text
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("Summary:", summary)
````

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**
A decoder-only model is designed specifically for autoregressive text generation, meaning it predicts the next word based on all preceding words in a continuous sequence. In conversational contexts, chat history and new prompts can simply be concatenated as context, allowing the model to naturally extend the dialogue one token at a time. Furthermore, using causal masking ensures the model only attends to past tokens, making the inference process highly efficient and well-suited for interactive, turn-by-turn conversations.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Artificial intelligence will"

print("--- Run 1 ---")
output1 = generator(prompt, max_new_tokens=30, do_sample=True, temperature=0.8)
print(output1[0]["generated_text"])

print("\n--- Run 2 ---")
output2 = generator(prompt, max_new_tokens=30, do_sample=True, temperature=0.8)
print(output2[0]["generated_text"])

print("\n--- Run 3 ---")
output3 = generator(prompt, max_new_tokens=30, do_sample=True, temperature=0.8)
print(output3[0]["generated_text"])
````

When do_sample=True is enabled, the model does not strictly pick the single most probable token (which is what Greedy Decoding does). Instead, it treats the output scores as a probability distribution and samples randomly from it. Without fixing the random seed using set_seed(), each generation run draws a different random token from that distribution, causing subsequent tokens and the entire resulting text to diverge.
